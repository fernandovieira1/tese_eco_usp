## 7. Conferir implementação contra PanelOLS

```python
from linearmodels.panel import PanelOLS
s=e.amostra(d,['pp_agricultura','expo_soja','expo_milho'],'verificacao')
a,rf,_=projetar(s,['pp_agricultura','expo_soja','expo_milho']);x=a[:,1:];b=np.linalg.solve(x.T@x,x.T@a[:,0]);u=a[:,0]-x@b
cv=covariancias(x*u[:,None],np.linalg.inv(x.T@x),s,rf+2)['DK2']
ref=PanelOLS.from_formula('pp_agricultura ~ expo_soja + expo_milho + EntityEffects + TimeEffects',s.set_index(['id','ano'])).fit(cov_type='kernel',kernel='bartlett',bandwidth=2,debiased=True)
assert np.allclose(b,ref.params,atol=1e-9)
assert np.allclose(cv,ref.cov,rtol=1e-6,atol=1e-10)
print('Coeficientes e covariância DK2 conferem com PanelOLS.')
```
