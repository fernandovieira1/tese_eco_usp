# Rodada 4 — dinâmica sob os controles que alteraram o padrão estático

```python
import sys
from pathlib import Path
sys.path.insert(0,str(Path.cwd().parent/'testes2'))
from metodos import *
from IPython.display import display
e=Experimento(Path.cwd());d=e.d
```

```python
for cultura in ['soja','milho']:
    for lag in [1,2,3]:d[f'expo_{cultura}_l{lag}']=d.groupby('id')['expo_'+cultura].shift(lag-1).where(d.ano.eq(d.groupby('id').ano.shift(lag-1)+lag-1))
x=[f'expo_{c}_l{l}' for c in ['soja','milho'] for l in [1,2,3]]
s0=d.dropna(subset=['pp_agricultura','pp_pastagem',*x])
```

## 1. Lags e tendências municipais — dois destinos

```python
s=s0[s0.ano.ge(2011)].copy() if False else s0.copy()
for uso in ['agricultura','pastagem']:
    ident='C01_'+uso
    r=e.linear(ident,'pp_'+uso,x,s,familia='dinamica',trend=True,bioma_ano=False)
    for cultura in ['soja','milho']:
        pesos={f'expo_{cultura}_l{l}':1 for l in [1,2,3]}
        e.contraste(ident,'Soma lags1–3 '+cultura,x,r['b'],r['cov']['DK2'],pesos,r['s'],'pp por 1DP persistente preço × 1DP aptidão','cumulativos')
    display(pd.DataFrame(e.mag).tail(2)[['id','contraste','estimativa','ic95_inf','ic95_sup','p']])
```

## 2. Lags e choques bioma×ano — dois destinos

```python
s=s0[s0.ano.ge(2011)].copy() if False else s0.copy()
for uso in ['agricultura','pastagem']:
    ident='C02_'+uso
    r=e.linear(ident,'pp_'+uso,x,s,familia='dinamica',trend=False,bioma_ano=True)
    for cultura in ['soja','milho']:
        pesos={f'expo_{cultura}_l{l}':1 for l in [1,2,3]}
        e.contraste(ident,'Soma lags1–3 '+cultura,x,r['b'],r['cov']['DK2'],pesos,r['s'],'pp por 1DP persistente preço × 1DP aptidão','cumulativos')
    display(pd.DataFrame(e.mag).tail(2)[['id','contraste','estimativa','ic95_inf','ic95_sup','p']])
```
