# OLS múltipla — composição de preços, aptidão e fiscalização

Modelos em complexidade crescente. M01C repete M01 na amostra de M02, para separar mudança de composição e adição de variável. Fiscalização é endógena, inclusive defasada.

**Inferência:** se a matriz two-way tiver autovalor negativo, aplica-se projeção espectral (autovalores negativos iguais a zero). SE bruto permanece exportado para auditoria; a correção é determinada pelo diagnóstico, nunca por significância. Sensibilidade municipal/HAC também é salva. A projeção não resolve poucos clusters nem ausência de identificação.

```python
from utilitarios import *
from IPython.display import display
d = ler()
print(f"{len(d)} linhas, {d.id.nunique()} municípios, {d.ano.min()}–{d.ano.max()}")
```

## 1. Preços e diferenças territoriais — dois destinos

```python
for uso in ['agricultura','pastagem']:
    ident='M01_'+uso; y='pp_'+uso
    colunas=[y,*['z_p_soja','z_p_milho','z_p_boi','z_a_soja','z_a_milho','bioma']] + []
    s=amostra(d,colunas,ident)
    formula=f'{y} ~ z_p_soja + z_p_milho + z_p_boi + z_a_soja + z_a_milho + C(bioma)'
    r,cov,cm=estimar(formula,s)
    display(exportar(ident,'Preços e diferenças territoriais',formula,s,r,cov,'OLS','bioma',y,cm=cm,alerta='seleção por disponibilidade da fiscalização; aptidão GAEZ não estritamente predeterminada'))
```

## 2. Mesma equação na amostra com fiscalização — dois destinos

```python
for uso in ['agricultura','pastagem']:
    ident='M01C_'+uso; y='pp_'+uso
    colunas=[y,*['z_p_soja','z_p_milho','z_p_boi','z_a_soja','z_a_milho','bioma']] + ['fiscal']
    s=amostra(d,colunas,ident)
    formula=f'{y} ~ z_p_soja + z_p_milho + z_p_boi + z_a_soja + z_a_milho + C(bioma)'
    r,cov,cm=estimar(formula,s)
    display(exportar(ident,'Mesma equação na amostra com fiscalização',formula,s,r,cov,'OLS','bioma',y,cm=cm,alerta='seleção por disponibilidade da fiscalização; aptidão GAEZ não estritamente predeterminada'))
```

## 3. Adição de autuações anteriores — dois destinos

```python
for uso in ['agricultura','pastagem']:
    ident='M02_'+uso; y='pp_'+uso
    colunas=[y,*['z_p_soja','z_p_milho','z_p_boi','z_a_soja','z_a_milho','bioma']] + ['fiscal']
    s=amostra(d,colunas,ident)
    formula=f'{y} ~ z_p_soja + z_p_milho + z_p_boi + z_a_soja + z_a_milho + C(bioma) + fiscal'
    r,cov,cm=estimar(formula,s)
    display(exportar(ident,'Adição de autuações anteriores',formula,s,r,cov,'OLS','bioma',y,cm=cm,alerta='seleção por disponibilidade da fiscalização; aptidão GAEZ não estritamente predeterminada'))
```
