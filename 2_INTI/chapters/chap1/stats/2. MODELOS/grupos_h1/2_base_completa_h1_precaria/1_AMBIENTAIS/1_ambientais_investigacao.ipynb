#### 1a_conversao

```python
df1a = pd.read_csv('1a_conversao.csv')
```

```python
df1a.head()
```

```python
# Transformar a coluna '0_ano' em datetime
df1a['0_ano'] = pd.to_datetime(df1a['0_ano'], format='%Y')

# Transformar as colunas '0_codigo_ibge' , '0_municipio' , '0_microrregiao' e '0_bioma_predominante' em categorias
df1a['0_codigo_ibge'] = df1a['0_codigo_ibge'].astype('category')
df1a['0_municipio'] = df1a['0_municipio'].astype('category')
df1a['0_microrregiao'] = df1a['0_microrregiao'].astype('category')
df1a['0_bioma_predominante'] = df1a['0_bioma_predominante'].astype('category')
```

```python
# Dados quantitativos
df1a.describe()
```

```python
# Dados qualitativos
df1a.describe(include=['category'])
```
