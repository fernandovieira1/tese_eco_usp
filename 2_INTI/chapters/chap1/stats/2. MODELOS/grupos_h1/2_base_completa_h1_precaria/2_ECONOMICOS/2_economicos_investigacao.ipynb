#### 2a_mercado_precos

```python
df2a = pd.read_csv('2a_mercado_precos.csv')
```

```python
df2a.head()
```

```python
# Transformar a coluna '0_ano' em datetime
df2a['0_ano'] = pd.to_datetime(df2a['0_ano'], format='%Y')

# Transformar as colunas '0_codigo_ibge' , '0_municipio' , '0_microrregiao' e '0_bioma_predominante' em categorias
df2a['0_codigo_ibge'] = df2a['0_codigo_ibge'].astype('category')
df2a['0_municipio'] = df2a['0_municipio'].astype('category')
df2a['0_microrregiao'] = df2a['0_microrregiao'].astype('category')
df2a['0_bioma_predominante'] = df2a['0_bioma_predominante'].astype('category')
```

```python
df2a.info()
```

```python
# Dados quantitativos
df2a.describe()
```

```python
# Dados qualitativos
df2a.describe(include=['category'])
```
