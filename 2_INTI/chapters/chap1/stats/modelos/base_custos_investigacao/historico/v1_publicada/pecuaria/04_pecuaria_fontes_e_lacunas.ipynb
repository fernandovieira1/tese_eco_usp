# Pecuária: referências de custos e lacunas

Os extratos oficiais transferidos e conferidos por SHA-256 contêm 196 registros de período em 25 abas: 132 de cria, 48 de recria e engorda conjuntamente e 16 de ciclo completo. Há 156 registros com COE/COT/CT numéricos. O formato longo mantém 588 campos monetários, dos quais 468 numéricos. Os três conceitos de uma mesma coluna não são observações independentes. A série estadual e as praças também não devem ser somadas como amostras independentes.

Todos os custos permanecem em **R$/arroba vendida nominal**. A produtividade modal em @vendida/ha está separada, sem multiplicação. Não se obtém fluxo de produção por hectare a partir de estoque de rebanho. Os valores não são despesas observadas de todos os produtores do município. Não houve municipalização, conversão para hectare, USD2024, modelagem ou incorporação à base3.

COE, COT e CT seguem os rótulos da fonte. Custo operacional, custo fixo/variável e custo irrecuperável são classificações distintas. Insumos, suplementação, serviços e manutenção são recorrentes; depreciação e remuneração do capital têm tratamento próprio; formação inicial de pastagem/instalações exige inventário de investimento e vida útil. Desmate não substitui custos operacionais. Evitar dupla contagem entre arrendamento e remuneração/depreciação da mesma terra/pastagem. Referência conceitual: https://cloud.cnpgc.embrapa.br/sac/2012/07/19/464-o-que-deve-ser-considerado-no-calculo-do-custo-de-producao-da-pecuaria-de-corte/ . Os extratos de totais não permitem classificar cada componente em fixo/variável ou irreversível.

As ausências são preservadas; 40 registros não têm custos. Guarantã do Norte e São José dos Quatro Marcos permanecem com identidade conflitante. A95 menciona algodão em várias abas: a inconsistência editorial permanece nas 752 células de notas, sem correção ou interpretação automática. Rótulos consolidados não constituem série trimestral completa; 2026 permanece fora da janela 2001–2024. Não se extrapolou para anos ausentes.
