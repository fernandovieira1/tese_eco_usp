# Rodada 4 — diferenças formais entre Amazônia e Cerrado

Classificação por bioma predominante. Pantanal excluído apenas destes contrastes; não recebe coeficiente sem suporte. As diferenças são testadas diretamente.

```python
import sys
from pathlib import Path
sys.path.insert(0,str(Path.cwd().parent/'testes2'))
from metodos import *
from IPython.display import display
e=Experimento(Path.cwd());d=e.d
```

```python
s=d[d.bioma.isin(['Amazônia','Cerrado'])].copy()
s['amazonia']=s.bioma.eq('Amazônia').astype(float)
s['expo_soja_amazonia']=s.expo_soja*s.amazonia;s['expo_milho_amazonia']=s.expo_milho*s.amazonia
print('Pares da grade excluídos por bioma:',len(d)-len(s))
```

## 1. Referência nos dois biomas

```python
x=['expo_soja','expo_milho']+(['expo_soja_amazonia','expo_milho_amazonia'] if False else [])
for u in ['agricultura','pastagem']:
    ident='H00_'+u
    r=e.linear(ident,'pp_'+u,x,s,familia='heterogeneidade',bioma_ano=False,joint=(x[2:] if False else x))
    if False:
        e.mag=[a for a in e.mag if a['id']!=ident]
        for cultura in ['soja','milho']:
            for bioma in ['Cerrado','Amazônia']:
                pesos={'expo_'+cultura:1,'expo_'+cultura+'_amazonia':int(bioma=='Amazônia')}
                e.contraste(ident,bioma+' slope '+cultura,x,r['b'],r['cov']['DK2'],pesos,r['s'],'pp por DP preço×DP aptidão','slopes_territoriais')
    display(pd.DataFrame(e.coef).tail(len(x))[['id','termo','coef','ic95_inf','ic95_sup','p','N']])
```

## 2. Slopes distintos por bioma

```python
x=['expo_soja','expo_milho']+(['expo_soja_amazonia','expo_milho_amazonia'] if True else [])
for u in ['agricultura','pastagem']:
    ident='H01_'+u
    r=e.linear(ident,'pp_'+u,x,s,familia='heterogeneidade',bioma_ano=False,joint=(x[2:] if True else x))
    if True:
        e.mag=[a for a in e.mag if a['id']!=ident]
        for cultura in ['soja','milho']:
            for bioma in ['Cerrado','Amazônia']:
                pesos={'expo_'+cultura:1,'expo_'+cultura+'_amazonia':int(bioma=='Amazônia')}
                e.contraste(ident,bioma+' slope '+cultura,x,r['b'],r['cov']['DK2'],pesos,r['s'],'pp por DP preço×DP aptidão','slopes_territoriais')
    display(pd.DataFrame(e.coef).tail(len(x))[['id','termo','coef','ic95_inf','ic95_sup','p','N']])
```
