```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Domínio de X (impacto ambiental)
x = np.linspace(0.1, 10, 400)  # começa em 0.1 para evitar divisão por zero ou log(0)

# Funções baseadas na proposta reformulada
# Predation: função quadrática com retorno decrescente
a, b = 2.5, 0.15
y_predatoria = a * x - b * x**2

# Extrativism: função logarítmica simples
y_extrativista = 2 * np.log(x + 1)

# Comensalism: log com deslocamento e ajuste de inclinação
y_comensal = 2.5 * np.log(0.8 * x + 2.5) + 0.3

# Mutualism: função quase linear com leve ganho marginal
y_mutualista = 0.5 * x + 0.8 * np.log(1 + x)

# Gráfico
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(x, y_predatoria, label='Predation', linestyle=':', linewidth=2, color='red')
ax.plot(x, y_extrativista, label='Extrativism', linestyle='--', linewidth=2, color='orange')
ax.plot(x, y_comensal, label='Comensalism', linestyle='-', linewidth=2, color='blue')
ax.plot(x, y_mutualista, label='Mutualism', linestyle='-.', linewidth=2, color='green')

# Linha de equilíbrio (x = y)
x_equilibrio = np.linspace(0, x.max() * 1.1, 100)
ax.plot(x_equilibrio, x_equilibrio, label='Balance', linestyle='--', color='gray')

# Estética do gráfico
ax.set_xlabel('Environmental impact (tCO₂e)')
ax.set_ylabel('Production (US$ billion)')
ax.set_ylim(0, max(y_predatoria.max(), y_comensal.max()) * 1.1)
ax.set_xlim(0, x.max() * 1.1)
fig.tight_layout()

# Ocultar os números dos eixos, manter grid
ax.tick_params(axis='both', which='both', labelbottom=False, labelleft=False)
ax.grid(True)

# Manter bordas visíveis
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(True)
ax.spines['left'].set_visible(True)

# Texto "x = y"
x_label_pos, y_label_pos = 6.4, 6
ax.annotate('X = Y',
            xy=(x_label_pos, y_label_pos),
            xytext=(x_label_pos + 0.3, y_label_pos + 0.3),
            fontsize=10,
            color='gray')

# Legenda
ax.legend(loc='lower right')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Domínio de X (impacto ambiental)
x = np.linspace(0, 10, 400)

# Funções de produção
y_predatoria = 2.8 * x - 0.3 * x**2
y_comensal =  0.06 * x**2 + 0.5 * x + 0.2
y_extrativista = 0.05 * x**2 + 0.69 * x + 0.10
y_mutualista = 0.78 * x - 0.09 * x**2

# Equilíbrio x = y
x_equilibrio = np.linspace(0, x.max() * 1.1, 100)
y_equilibrio = x_equilibrio

# Cálculo automático dos pontos x^* e x̄
x_star = 0.78 / 0.09  # raiz de y_mutualista = 0
x_bar = 2.8 / 0.3     # raiz de y_predatoria = 0

# Gráfico
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(x, y_predatoria, label='Predation', linestyle=':', linewidth=2, color='red')
ax.plot(x, y_extrativista, label='Extrativism', linestyle='--', linewidth=2, color='orange')
ax.plot(x, y_comensal, label='Comensalism', linestyle='-', linewidth=2, color='royalblue')
ax.plot(x, y_mutualista, label='Mutualism', linestyle='-.', linewidth=2, color='green')
ax.plot(x_equilibrio, y_equilibrio, label='Balance', linestyle='-', linewidth=2, color='darkgray')

# Estética
ax.set_xlabel('Environmental impact (tCO₂e)', loc='left', labelpad=5)
ax.set_ylabel('Production (US$ billion)')
ax.set_ylim(0, max(y_predatoria.max(), y_comensal.max(), y_extrativista.max(), y_mutualista.max()) * 1.1)
ax.set_xlim(0, x.max() * 1.1)
fig.tight_layout()
ax.tick_params(axis='both', which='both', labelbottom=False, labelleft=False)
ax.grid(True, color='whitesmoke')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Texto "x = y"
ax.annotate('X = Y',
            xy=(10, 10.7),
            xytext=(10.3, 11),
            fontsize=10,
            color='gray')

# Legenda
ax.legend(loc='upper left')

# Ponto x de comparação
x_sample = 5
Pc_x = 0.06 * x_sample**2 + 0.5 * x_sample + 0.2
Pe_x = 0.05 * x_sample**2 + 0.69 * x_sample + 0.10

# Pontos P_c(x) e P_e(x)
ax.plot(x_sample, Pc_x, 'o', color='royalblue')
ax.annotate(r'$P_c(x)$', xy=(x_sample, Pc_x), xytext=(x_sample + 0.2, Pc_x - 0.5), color='royalblue')
ax.plot(x_sample, Pe_x, 'o', color='orange')
ax.annotate(r'$P_e(x)$', xy=(x_sample, Pe_x), xytext=(x_sample - 0.6, Pe_x + 0.1), color='orange')

# Linhas verticais para x, x^*, e x̄
ax.axvline(x=x_sample, linestyle='-', color='black', linewidth=1)
ax.annotate(r'$x´$', xy=(x_sample, 0), xytext=(x_sample - 0.1, -0.5))

ax.axvline(x=x_star, linestyle='-', color='black', linewidth=1)
ax.annotate(r'$x^*$', xy=(x_star, 0), xytext=(x_star - 0.1, -0.5))

ax.axvline(x=x_bar, linestyle='-', color='black', linewidth=1)
ax.annotate(r'$\bar{x}$', xy=(x_bar, 0), xytext=(x_bar - 0.1, -0.5))

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Domínio de X (impacto ambiental)
x = np.linspace(0, 10, 400)

# Funções de produção
y_predatoria = 2.8 * x - 0.3 * x**2
y_comensal =  0.06 * x**2 + 0.5 * x + 0.2
y_extrativista = 0.05 * x**2 + 0.69 * x + 0.10
y_mutualista = 0.78 * x - 0.09 * x**2

# Equilíbrio x = y
x_equilibrio = np.linspace(0, x.max() * 1.1, 100)
y_equilibrio = x_equilibrio

# Cálculo automático dos pontos x^* e x̄
x_star = 0.78 / 0.09  # raiz de y_mutualista = 0
x_bar = 2.8 / 0.3     # raiz de y_predatoria = 0

# Filtro para sombrear a área da vantagem comensalista (𝓒)
mask = (x >= 0) & (x <= x_star)
x_fill = x[mask]
yc_fill = y_comensal[mask]
ye_fill = y_extrativista[mask]

# Gráfico
fig, ax = plt.subplots(figsize=(10, 6))

# Curvas principais
ax.plot(x, y_predatoria, label='Predation', linestyle=':', linewidth=2, color='red')
ax.plot(x, y_extrativista, label='Extrativism', linestyle='--', linewidth=2, color='orange')
ax.plot(x, y_comensal, label='Comensalism', linestyle='-', linewidth=2, color='royalblue')
ax.plot(x, y_mutualista, label='Mutualism', linestyle='-.', linewidth=2, color='green')
ax.plot(x_equilibrio, y_equilibrio, label='Balance', linestyle='-', linewidth=2, color='darkgray')

# Sombreamento da área 𝓒
ax.fill_between(x_fill, yc_fill, ye_fill, where=ye_fill > yc_fill, interpolate=True,
                color='skyblue', alpha=0.4, label=r'$\mathcal{C}$ (commensal advantage)')

# Estética
ax.set_xlabel('Environmental impact (tCO₂e)', loc='left', labelpad=5)
ax.set_ylabel('Production (US$ billion)')
ax.set_ylim(0, max(y_predatoria.max(), y_comensal.max(), y_extrativista.max(), y_mutualista.max()) * 1.1)
ax.set_xlim(0, x.max() * 1.1)
fig.tight_layout()
ax.tick_params(axis='both', which='both', labelbottom=False, labelleft=False)
ax.grid(True, color='whitesmoke')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Texto "X = Y"
ax.annotate('X = Y', xy=(10, 10.7), xytext=(10.3, 11), fontsize=10, color='gray')

# Legenda
ax.legend(loc='upper left')

# Pontos P_c(x) e P_e(x)
x_sample = 5
Pc_x = 0.06 * x_sample**2 + 0.5 * x_sample + 0.2
Pe_x = 0.05 * x_sample**2 + 0.69 * x_sample + 0.10
ax.plot(x_sample, Pc_x, 'o', color='royalblue')
ax.annotate(r'$P_c(x)$', xy=(x_sample, Pc_x), xytext=(x_sample + 0.2, Pc_x - 0.5), color='royalblue')
ax.plot(x_sample, Pe_x, 'o', color='orange')
ax.annotate(r'$P_e(x)$', xy=(x_sample, Pe_x), xytext=(x_sample - 0.6, Pe_x + 0.1), color='orange')

# Linhas verticais
ax.axvline(x=x_sample, linestyle='-', color='black', linewidth=1)
ax.annotate(r"$x'$", xy=(x_sample, 0), xytext=(x_sample - 0.1, -0.5))
ax.axvline(x=x_star, linestyle='-', color='black', linewidth=1)
ax.annotate(r'$x^*$', xy=(x_star, 0), xytext=(x_star - 0.1, -0.5))
ax.axvline(x=x_bar, linestyle='-', color='black', linewidth=1)
ax.annotate(r'$\bar{x}$', xy=(x_bar, 0), xytext=(x_bar - 0.1, -0.5))

plt.show()
```
