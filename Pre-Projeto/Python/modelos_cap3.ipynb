```python
from matplotlib.table import Table
import textwrap

import matplotlib.pyplot as plt

# Function to wrap text in each cell
def wrap_text(text, width=30):
    return '\n'.join(textwrap.wrap(text, width=width))

# Create figure with appropriate size
fig, ax = plt.subplots(figsize=(16, 12))
ax.axis('off')

# Prepare the data with wrapped text
wrapped_data = extended_df.copy()
for col in wrapped_data.columns:
    wrapped_data[col] = wrapped_data[col].apply(lambda x: wrap_text(x))

# Calculate number of rows and columns
nrows, ncols = wrapped_data.shape

# Create the table
table = Table(ax, bbox=[0, 0, 1, 1])

# Add column headers
for i, col in enumerate(wrapped_data.columns):
    wrapped_header = wrap_text(col, width=20)
    table.add_cell(0, i, 1, 1, text=wrapped_header, 
                  loc='center', edgecolor='black', facecolor='lightgrey')

# Add data cells
for i in range(nrows):
    for j in range(ncols):
        cell_text = wrapped_data.iloc[i, j]
        table.add_cell(i+1, j, 1, 1, text=cell_text, 
                      loc='center', edgecolor='black')

# Add table to plot
ax.add_table(table)

# Calculate the best height for the figure based on content
max_lines_per_cell = 1
for (i, j), cell in table._cells.items():
    if cell.get_text():
        lines = cell.get_text().count('\n') + 1
        max_lines_per_cell = max(max_lines_per_cell, lines)

fig.set_figheight(nrows * max_lines_per_cell * 0.4 + 2)

# Set title
plt.title('Modelos Econométricos para Análise', fontsize=16, pad=20)
plt.tight_layout()
# Calculate the best height for the figure based on content
max_lines_per_cell = 1
for (i, j), cell in table._cells.items():
    if cell.get_text():
        lines = cell.get_text().count('\n') + 1
        max_lines_per_cell = max(max_lines_per_cell, lines)

fig.set_figheight(nrows * max_lines_per_cell * 0.4 + 2)
# Save to PNG with high DPI for better quality
plt.savefig('extended_df_formatted.png', dpi=300, bbox_inches='tight')
plt.close()

print("Table exported as 'extended_df_formatted.png'")
```
