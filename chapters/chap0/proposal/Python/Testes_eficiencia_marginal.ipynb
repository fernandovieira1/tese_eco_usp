```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros base do modelo
A = 1  # produtividade total dos fatores
alpha = 0.4  # elasticidade do capital
beta = 0.4   # elasticidade do trabalho
gamma = 0.2  # elasticidade do meio ambiente (comensalismo)

# Preços dos fatores
r = 1.0   # custo do capital
w = 1.0   # custo do trabalho
p_range = np.linspace(0.1, 1.5, 100)  # custo ambiental (vai de incentivo até penalidade)

# Níveis fixos de capital e trabalho
K = 10
L = 10

# Função de produção para diferentes valores de E e custo ambiental p
E = np.linspace(1, 20, 100)
Y = lambda E_val: A * (K**alpha) * (L**beta) * (E_val**gamma)

# Cálculo da produção e da razão benefício/custo ambiental marginal (dY/dE / p)
Y_vals = Y(E)
dYdE = gamma * A * (K**alpha) * (L**beta) * (E**(gamma - 1))

# Simulando benefício/custo marginal ecológico para diferentes p
efficiency_ratios = np.array([dYdE / p for p in p_range])

# Gerar o gráfico
fig, ax = plt.subplots(figsize=(10, 6))
for i in range(0, len(p_range), 20):
    ax.plot(E, efficiency_ratios[i], label=f'p = {p_range[i]:.2f}')

ax.set_title('Eficiência marginal do uso ambiental (dY/dE / p)')
ax.set_xlabel('Uso ambiental (E)')
ax.set_ylabel('Eficiência marginal ajustada ao custo')
ax.axhline(1, color='gray', linestyle='--', linewidth=1, label='limite: ganho = custo')
ax.legend()
plt.grid(True)


plt.show()
```
